For points $(x_1, y_1, z_1)$ and $(x_2, y_2, z_2)$:

$$d = \sqrt{(x_1-x_2)^2 + (y_1-y_2)^2 + (z_1-z_2)^2}$$


#### TypeError

````python
def distance(A, B):
    """A and B are lists/tuples with the x, y, z coordinates of two points in 3-space."""
    try:
        return ((A[0]-B[0])**2 + (A[1]-B[1])**2 + (A[2]-B[2])**2) ** 0.5
    except TypeError:
        print("check validity of the inputs!")

distance([0, 0, 0], [1, 1, "a"])
````

<h2 style="color:#1e3a8a;">NumPy</h2>

- Vectorization of the code&hellip;
- NumPy data structures take up less space than plain Python lists
    - A Python list stores *pointers* to separate Python objects scattered in memory, each with its own type info and overhead. A NumPy array stores raw values of one fixed type in one contiguous block of memory — so operations run as tight, compiled C loops instead of Python's interpreted per-element loop, and the CPU can read that contiguous memory far more efficiently (better cache usage).
- NumPy contains many optimized functions for faster calculations
- It serves as the foundation for many other popular libraries, such as Pandas and Matplotlib

### NumPy arrays

- NumPy's main object is the homogeneous multidimensional array -- every element is guaranteed to be the same fixed type and size.
- It is a table of elements (usually numbers), all of the same type, indexed by a tuple of positive integers.
- In NumPy, dimensions are called *axes*; the number of axes is the *rank* (e.g. `np.array([[1,2],[3,4]])` has rank 2).
- Scalar, vector, matrix, tensor:

    | Term | Rank | Shape example | Code example |
    |---|---|---|---|
    | Scalar | 0 | `()` | `np.array(5)` -- a single number |
    | Vector | 1 | `(3,)` | `np.array([1, 2, 3])` -- a 1D list of numbers |
    | Matrix | 2 | `(2, 2)` | `np.array([[1, 2], [3, 4]])` -- rows and columns |
    | Tensor | 3+ | `(2, 2, 2)` | `np.array([[[1,2],[3,4]], [[5,6],[7,8]]])` -- a "stack" of matrices, or any array with 3+ axes |

    (A scalar, vector, and matrix are technically all tensors too -- rank 0, 1, and 2. People usually
only start saying "tensor" once you're at rank 3 or higher, where you can't picture it as a flat table.)
- NumPy's array class is called `ndarray` (alias `array`).

#### Broadcasting

NumPy lets you combine arrays of *different* shapes without writing a loop, by automatically
"stretching" the smaller one to match &mdash; this is called **broadcasting**, and it's a big part of
why NumPy code is both shorter and faster than plain Python.

````python
prices = np.array([100, 200, 150])

print(prices + 10)          # a scalar is broadcast to every element

matrix = np.array([[1, 2, 3],
                    [4, 5, 6]])
row_adjustment = np.array([10, 20, 30])

print(matrix + row_adjustment)   # the 1D array is broadcast across every row
````

### Polyfit

A quick, practical example of what a built-in NumPy function can do with an array of data -- shown here
purely for the *mechanics* (array in, coefficients out), not to teach model-fitting properly. Given some
x/y data, `polyfit` finds the coefficients of the best-fitting line (or curve) by least squares. We will
come back to actually fitting, interpreting, and evaluating models with real care starting Day 3
(regression & causality) and Day 4 (ML evaluation) -- for now, just watch what the function does.


````python
# creating random observations
err = np.random.random(100)      # 100 random values between 0 and 1
x = np.linspace(0, 1, 100)       # 100 evenly spaced values from 0 to 1
y = 3 + 2 * x + err
````

````python
# don't worry, we'll cover plotting properly in the next session :)
import matplotlib.pyplot as plt
plt.plot(x, y, ".")
plt.show()
````

We use `numpy.polyfit(x, y, deg)` for a least-squares polynomial fit
(docs: https://numpy.org/doc/stable/reference/generated/numpy.polyfit.html).

The output is an array of coefficients $\beta$, from degree 0 to $p$: $\sum_{i=0}^{p} \beta_i \, x^{i}$

````python
fitted = np.polyfit(x, y, 1)   # degree 1 = a straight line: [slope, intercept]
print(fitted)
````

````python
y_hat = x * fitted[0] + fitted[1]
````

````python
plt.plot(x, y, ".")
plt.plot(x, y_hat)   # no marker specified -> defaults to a continuous line
plt.show()
````

**Extrapolation**

````python
# at x = 2, where y = x * slope + intercept
2 * fitted[0] + fitted[1]
````

That's it for now -- deliberately. We're stopping *before* asking "is this a good fit?",
"what do these coefficients actually mean?", or "should I trust this line outside my data?" (notice the
extrapolation above quietly assumes the same straight-line relationship holds *beyond* the x range we
actually observed -- exactly the kind of assumption we'll learn to question properly later). Those
questions deserve real treatment, not a quick NumPy demo -- that's exactly what regression, causality,
and model evaluation are all about later in the course.
