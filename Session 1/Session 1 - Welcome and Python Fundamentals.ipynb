<span style="color:#b45309">**Write a function to return the distance between two points in 3D Euclidean space:**</span>

For points $(x_1, y_1, z_1)$ and $(x_2, y_2, z_2)$:

$$d = \sqrt{(x_1-x_2)^2 + (y_1-y_2)^2 + (z_1-z_2)^2}$$


````python
def distance(A, B):
    """A and B are lists/tuples with the x, y, z coordinates of two points in 3-space."""
    return ((A[0]-B[0])**2 + (A[1]-B[1])**2 + (A[2]-B[2])**2) ** 0.5
````

````python
distance([0, 0, 0], [1, 1, 1])
````

#### Broadcasting 🟢

NumPy lets you combine arrays of *different* shapes without writing a loop, by automatically
"stretching" the smaller one to match &mdash; this is called **broadcasting**, and it's a big part of
why NumPy code is both shorter and faster than plain Python.

````python
prices = np.array([100, 200, 150])

print(prices + 10)          # a scalar is broadcast to every element

matrix = np.array([[1, 2, 3],
                    [4, 5, 6]])
row_adjustment = np.array([10, 20, 30])

print(matrix + row_adjustment)   # the 1D array is broadcast across every row
````

### Polyfit 🟡

````python
# creating random observations
err = np.random.random(100)      # 100 random values between 0 and 1
x = np.linspace(0, 1, 100)       # 100 evenly spaced values from 0 to 1
y = 3 + 2 * x + err
````

````python
# don't worry, we'll cover plotting properly in the next session :)
import matplotlib.pyplot as plt
plt.plot(x, y, ".")
plt.show()
````

We use `numpy.polyfit(x, y, deg)` for a least-squares polynomial fit
(docs: https://numpy.org/doc/stable/reference/generated/numpy.polyfit.html).

The output is an array of coefficients $\beta$, from degree 0 to $p$: $\sum_{i=0}^{p} \beta_i \, x^{i}$

````python
fitted = np.polyfit(x, y, 1)   # degree 1 = a straight line: [slope, intercept]
print(fitted)
````

````python
y_hat = x * fitted[0] + fitted[1]
````

````python
plt.plot(x, y, ".")
plt.plot(x, y_hat)   # no marker specified -> defaults to a continuous line
plt.show()
````

**Extrapolation**

````python
# at x = 2, where y = x * slope + intercept
2 * fitted[0] + fitted[1]
````

#### Class exercise 2: calculate the $R^2$ of the fitted line

$$R^2 = 1 - \frac{\sum_i (y_i - \hat y_i)^2}{\sum_i (y_i - \bar y)^2}$$

````python
def r_squared(y, y_hat):
    """
    Parameters:
        y (array-like): actual y values
        y_hat (array-like): predicted y values from the model
    Returns:
        float: R-squared value
    """
    ss_res = np.sum((y - y_hat) ** 2)        # sum of squared residuals
    ss_tot = np.sum((y - np.mean(y)) ** 2)   # total sum of squares
    return 1 - (ss_res / ss_tot)

r_squared(y, y_hat)
````

#### Class exercise 3: does $R^2$ increase with a degree-2 fit? 🟢
*(WS2025's version stopped right before finishing this &mdash; completed here.)*

````python
fitted2 = np.polyfit(x, y, 2)
y_hat2 = fitted2[0] * x**2 + fitted2[1] * x + fitted2[2]

print("Degree 1 R^2:", r_squared(y, y_hat))
print("Degree 2 R^2:", r_squared(y, y_hat2))

# A higher-degree polynomial will (almost) always fit the observed data at least as well --
# that's not the same as being a *better model*. We'll return to this exact tension (overfitting,
# in-sample vs. out-of-sample fit) when we cover the bias-variance trade-off later in the course.
````
